```python
import pandas as pd
import numpy as np
from typing import Dict, Optional, List, Tuple, Any
from pathlib import Path

df = pd.read_csv("/Users/mohamedkamel/UN_repo/policy-pulse/data/resolution_table.csv")
```

```python
df.head()
```

```python
country_options = []
excluded_cols = [
    "undl_id", "date", "session", "resolution", "draft", "committee_report", 
    "meeting", "title", "agenda_title", "subjects", "total_yes", "total_no",
    "total_abstentions", "total_non_voting", "total_ms", "undl_link"
]
for c in df.columns:
    if c not in excluded_cols:
        country_options.append(c)
country_options.sort()
```

```python
MAP = []
for country in country_options:
    min_date = '2026'
    l = df.query(f"{country} in ['Y','N','A']")['date'].to_list()
    for date in l:
        if date < min_date:
            min_date = date
    MAP.append({"country": country, "min_date": min_date})
    
```
